```python
#3

bimestre = 1

soma_notas = 0.0

while bimestre <= 4:

    nota = float(input(f"Digite a nota do {bimestre} bimestre: "))

    soma_notas += nota

    bimestre += 1

media = soma_notas / 4

if media >= 7.0:
    resultado = "Aprovado"
elif media >= 5.0:
    resultado = "Recuperação"
else:
    resultado = "Reprovado"

print(f"Média final: {media:.2}")
print(f"Situação do aluno: {resultado}")
```
