```
# Fazer um jogo de adivinhação

# O usuario ira adivinhar um número de 0 até 10.

# regras

# O usúario terá 03 chances de adivinhar
# Informar se o palpite do usúario é maior ou menor que o número.
# Imprimir mensagem de está quente, morno ou frio.



```

```
import random

def jogo_adivinhacao():
    numero_secreto = random.randint(0, 10)
    chances = 3
    tentativas_feitas = 0

    print("Bem-vindo ao jogo de adivinhação!")
    print("Tente adivinhar um número entre 0 e 10. Você tem 3 chances.")

    while tentativas_feitas < chances:
        palpite = int(input(f"\nTentativa {tentativas_feitas + 1} de {chances}. Qual é o seu palpite? "))
        tentativas_feitas += 1


        if palpite == numero_secreto:
            print("Parabéns! Você acertou o número!")

            tentativas_feitas = chances


        else:

            diferenca = abs(palpite - numero_secreto)
            if diferenca == 0:

            elif diferenca <= 1:
                print("Está quente! Muito perto do número secreto.")
            elif diferenca <= 3:
                print("Está morno! Chegando perto, mas ainda não é esse.")
            else:
                print("Está frio! Longe do número secreto.")


            if tentativas_feitas < chances:
                if palpite > numero_secreto:
                    print("Dica: O seu palpite é MAIOR que o número secreto.")
                else:
                    print("Dica: O seu palpite é MENOR que o número secreto.")


    if palpite != numero_secreto:
        print(f"\nSuas chances acabaram! O número secreto era {numero_secreto}.")

jogo_adivinhacao()
```
