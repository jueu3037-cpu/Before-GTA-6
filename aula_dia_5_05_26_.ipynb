```python
# 1 - Escrever um programa que o usuario digita um número e o programa imprimi até o número digitado.

# Solicita a palavra ao usuário
palavra = input("Digite uma palavra: ")

# Define as vogais (incluindo maiúsculas para garantir)
vogais = "aeiouAEIOU"
contador = 0

# Percorre cada letra na palavra digitada
for letra in palavra:
    # Verifica se a letra está na string de vogais
    if letra in vogais:
        contador += 1

# Exibe o resultado
print(f"A palavra '{palavra}' tem {contador} vogais.")
```

```python
# 2 - Escrever um programa que o usuario digita um número e o programa imprimi o número digitado até o zero (contagem regressiva).

# 1. Pede ao usuário para digitar um número inteiro

numero = int(input("Digite um número inteiro positivo para a contagem: "))

# 2. Faz a contagem regressiva de 'numero' até 0
# range(início, fim_exclusivo, passo)
# Começa em 'numero', vai até -1 para incluir o 0, e diminui de 1 em 1
for i in range(numero, -1, -1):
    print(i)

```

```python
# 3 - Escrever um programa que o usuario digita um número e o programa imprimi os números pares.

# Solicita o número ao usuário
n = int(input("Digite um número inteiro: "))

print(f"Números pares entre 0 e {n}:")

# range(2, n+1, 2) começa em 2, vai até n, pulando de 2 em 2
for i in range(2, n + 1, 2):
    print(i)
```

```python
# 4 - Escrever um programa que o usuario digita um número e o programa imprimi os números impares.

# Solicita o número ao usuário
n = int(input("Digite um número inteiro: "))

print(f"Números ímpares de 1 até {n}:")

# Loop começando em 1, até n+1, pulando de 2 em 2
for i in range(1, n + 1, 2):
    print(i)
```

```python
# 5 - Escrever um programa que o usuario digita um número e programa soma dos números até o número digitado.

# Solicita o número ao usuário
numero = int(input("Digite um número inteiro positivo: "))

# Calcula a soma dos números até o número digitado
soma = 0
for i in range(1, numero + 1):
    soma += i

# Exibe o resultado
print(f"A soma de 1 até {numero} é: {soma}")

```

```python
# 6 - Escrever um programa que o usuario digita uma palavra e o programa conta as letras da palavra.

# O programa pede para o usuário digitar algo
palavra = input("Digite uma palavra: ")

# A função len() conta quantos caracteres existem na variável
quantidade = len(palavra)

# Exibe o resultado
print(f"A palavra '{palavra}' tem {quantidade} letras.")


```

```python
# 7 - Escrever um programa que o usuario digita uma palavra e o programa conta as vogais da palavra.

# Solicita a palavra ao usuário
palavra = input("Digite uma palavra: ").lower()

# Define o que é uma vogal
vogais = "aeiouáéíóúâêîôûãõ"
contador = 0

# Percorre a palavra e conta as vogais
for letra in palavra:
    if letra in vogais:
        contador += 1

print(f"A palavra tem {contador} vogais.")


```

```python
# 8 - Escrever um programa que o usuario digita uma palavra e o programa imprimi a palavra invertida.

# Solicita a palavra ao usuário
palavra = input("Digite uma palavra: ")

# Inverte a palavra usando fatiamento [início:fim:passo]
# O passo -1 faz a leitura de trás para frente
palavra_invertida = palavra[::-1]

# Exibe o resultado
print(f"A palavra invertida é: {palavra_invertida}")


```

```python
# 9 - Escrever um programa que o usuario digita 04 números e o programa imprimi o maior.

# Solicita ao usuário que digite os 4 números
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
num3 = float(input("Digite o terceiro número: "))
num4 = float(input("Digite o quarto número: "))

# Utiliza a função max() para encontrar o maior entre os quatro
maior = max(num1, num2, num3, num4)

# Imprime o resultado
print(f"O maior número digitado é: {maior}")

```
