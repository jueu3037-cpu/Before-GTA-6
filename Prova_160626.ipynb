```python
# 2

numero = int(input("Digite um número inteiro positivo: "))

soma = 0

for i in range(1, numero + 1):
    soma += i

print(f"A soma de todos os números de 1 até {numero} é: {soma}")
```

```python
# 3

numero = int(input("Digite um número inteiro para ver a sua tabuada: "))

print(f"\nTabuada do {numero}:")

for i in range(1, 11):
    resultado = numero * i
    print(f"{numero} x {i:2d} = {resultado}")
```

```python
# 5

cedulas = [100, 50, 20, 10, 5, 2, 1]

try:
    valor_saque = int(input("Digite o valor do saque (R$): "))

    if valor_saque <= 0:
        print("Valor inválido. Digite um valor maior que zero.")
    else:
        valor_restante = valor_saque
        print(f"\nPara sacar R$ {valor_saque}, o caixa entregará a menor quantidade de notas:")

        for nota in cedulas:
            quantidade = valor_restante // nota

            if quantidade > 0:
                print(f"{quantidade} nota(s) de R$ {nota}")
                valor_restante %= nota

except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```
