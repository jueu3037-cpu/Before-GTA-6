```
#Escrever um programa que recebe 04 notas referente a cada bimestrte e calcula a média das notas e imprimi a média do aluno
# E se foi ele foi aprovado, recuperação ou reprovado.

# regras

# utilizar while
#definir uma variaval para o bimestre e uma váriavel pra nota e uma variavel para o resultado.

bimestre = 1

soma_notas = 0.0

while bimestre <= 4:

    nota = float(input(f"Digite a nota do {bimestre} bimestre: "))

    soma_notas += nota

    bimestre += 1

media = soma_notas / 4

if media >= 7.0:
    resultado = "Aprovado"
elif media >= 5.0:
    resultado = "Recuperação"
else:
    resultado = "Reprovado"

print(f"Média final: {media:.2}")
print(f"Situação do aluno: {resultado}")


```
